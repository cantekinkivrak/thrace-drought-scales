### 3 · Landsat C2 L2 collection

```python
SR_SCALE,SR_OFF=0.0000275,-0.2; ST_SCALE,ST_OFF=0.00341802,149.0
SD={'L5':('1984-03-01','2012-05-05'),'L7':('1999-05-28','2022-04-06'),'L8':('2013-03-18','2025-12-31'),'L9':('2021-10-31','2025-12-31')}
CFG={'L5':('LANDSAT/LT05/C02/T1_L2',['SR_B1','SR_B2','SR_B3','SR_B4','SR_B5','SR_B7','ST_B6','QA_PIXEL']),
     'L7':('LANDSAT/LE07/C02/T1_L2',['SR_B1','SR_B2','SR_B3','SR_B4','SR_B5','SR_B7','ST_B6','QA_PIXEL']),
     'L8':('LANDSAT/LC08/C02/T1_L2',['SR_B2','SR_B3','SR_B4','SR_B5','SR_B6','SR_B7','ST_B10','QA_PIXEL']),
     'L9':('LANDSAT/LC09/C02/T1_L2',['SR_B2','SR_B3','SR_B4','SR_B5','SR_B6','SR_B7','ST_B10','QA_PIXEL'])}
COMMON=['blue','green','red','nir','swir1','swir2','thermal','QA_PIXEL']; OPT=['blue','green','red','nir','swir1','swir2']
ROY_S={'blue':0.8474,'green':0.8483,'red':0.9047,'nir':0.8462,'swir1':0.8937,'swir2':0.9071}
ROY_I={'blue':0.0003,'green':0.0088,'red':0.0061,'nir':0.0412,'swir1':0.0254,'swir2':0.0172}
def mask_qa(img):
    qa=img.select('QA_PIXEL')
    bad=(qa.bitwiseAnd(1<<1).neq(0).Or(qa.bitwiseAnd(1<<2).neq(0)).Or(qa.bitwiseAnd(1<<3).neq(0)).Or(qa.bitwiseAnd(1<<4).neq(0)).Or(qa.bitwiseAnd(1<<5).neq(0)))
    return img.updateMask(bad.Not())
def prep(s):
    cid,bands=CFG[s]; s0,s1=SD[s]
    lo=f'{max(int(s0[:4]),BASELINE[0])}-01-01'; hi=f'{min(int(s1[:4]),BASELINE[1])}-12-31'
    c=ee.ImageCollection(cid).filterBounds(roi).filterDate(lo,hi).map(mask_qa)
    def _p(img):
        img=img.select(bands,COMMON); opt=img.select(OPT).multiply(SR_SCALE).add(SR_OFF)
        if HARMONIZE and s in ('L5','L7'): opt=opt.multiply(ee.Image([ROY_S[b] for b in OPT])).add(ee.Image([ROY_I[b] for b in OPT]))
        lst=img.select('thermal').multiply(ST_SCALE).add(ST_OFF).subtract(273.15).rename('LST')
        ndvi=opt.normalizedDifference(['nir','red']).rename('NDVI')
        return ndvi.addBands(lst).copyProperties(img,['system:time_start'])
    return c.map(_p)
landsat=prep(SENSORS[0])
for s in SENSORS[1:]: landsat=landsat.merge(prep(s))
landsat=landsat.select(['NDVI','LST'])
print('Collection built.')
```

### 5 · Gridded VHI function + August climatology

```python
def monthly_image(year, month):
    a=ee.Date.fromYMD(year,month,1); b=a.advance(1,'month'); sub=landsat.filterDate(a,b)
    return sub.select('NDVI').max().rename('NDVI').addBands(sub.select('LST').mean().rename('LST'))

def month_clim(month):
    ic=ee.ImageCollection([monthly_image(y,month) for y in range(BASELINE[0],BASELINE[1]+1)])
    return (ic.select('NDVI').min().rename('nmin').addBands(ic.select('NDVI').max().rename('nmax'))
            .addBands(ic.select('LST').min().rename('tmin')).addBands(ic.select('LST').max().rename('tmax')))

CLIM = month_clim(TARGET_MONTH)   # computed once for the target month

def vhi_image(year, month=TARGET_MONTH, clim=None):
    c = clim if clim is not None else month_clim(month)
    img=monthly_image(year,month)
    vci=img.select('NDVI').subtract(c.select('nmin')).divide(c.select('nmax').subtract(c.select('nmin'))).multiply(100)
    tci=c.select('tmax').subtract(img.select('LST')).divide(c.select('tmax').subtract(c.select('tmin'))).multiply(100)
    vhi=vci.multiply(ALPHA).add(tci.multiply(1-ALPHA)).rename('VHI').clamp(0,100)
    return vhi.updateMask(cropland).clip(roi)
print('VHI image function ready.')
```

### 6 · Publication figure (saves PNG + PDF)

```python
# =====================================================================
# PUBLICATION MAPS FIGURE — add as a cell to VHI_Drought_Maps_Trakya.ipynb
# (run after §4 defines vhi_image / CLIM). Produces a 300-dpi multi-panel
# PNG+PDF with per-panel SPEI-3 annotation and a shared colorbar.
# =====================================================================
import matplotlib.pyplot as plt, matplotlib as mpl, urllib.request, numpy as np, PIL.Image as PImage
mpl.rcParams.update({'font.size':9,'figure.dpi':300,'savefig.dpi':300})

# Regional growing-season SPEI-3 for each mapped year (from the SPI/SPEI results)
SPEI3 = {2001:-0.71, 2007:-0.97, 2012:-0.83, 2024:-1.77, 2014:+1.54}
PANELS = [2001, 2007, 2012, 2024, 2014]        # drought years + wet contrast
PALETTE = ['a50026','d73027','f46d43','fee08b','a6d96a','1a9850']

def thumb(img, px=520):
    url = img.getThumbURL({'min':0,'max':100,'region':roi,'dimensions':px,'palette':PALETTE})
    return np.array(PImage.open(urllib.request.urlopen(url)))

fig, axes = plt.subplots(1, len(PANELS), figsize=(2.5*len(PANELS), 3.0))
for ax, y in zip(axes, PANELS):
    ax.imshow(thumb(vhi_image(y, clim=CLIM))); ax.axis('off')
    cat = 'wet' if SPEI3[y] > 0.5 else ('severe drought' if SPEI3[y] < -1 else 'drought')
    ax.set_title(f'{y}   (SPEI-3 = {SPEI3[y]:+.2f})\n{cat}', fontsize=8.5)

# shared horizontal colorbar
sm = mpl.cm.ScalarMappable(norm=mpl.colors.Normalize(0,100),
       cmap=mpl.colors.LinearSegmentedColormap.from_list('vhi', ['#'+c for c in PALETTE]))
cax = fig.add_axes([0.30, 0.06, 0.40, 0.03])
cb = fig.colorbar(sm, cax=cax, orientation='horizontal'); cb.set_label('VHI (0 = stressed, 100 = healthy)', fontsize=8)
fig.suptitle('Landsat Vegetation Health Index — Thrace rainfed cropland (August, α = 0.5)',
             y=0.98, fontsize=10, fontweight='bold')
fig.text(0.5, 0.005, 'CORINE-211 (non-irrigated arable) mask. İpsala valley (irrigated rice) not separated here.',
         ha='center', fontsize=6.5, color='#7f8c8d', style='italic')
plt.subplots_adjust(bottom=0.18, top=0.82, wspace=0.05)
fig.savefig('Fig_VHI_drought_maps.png', bbox_inches='tight')
fig.savefig('Fig_VHI_drought_maps.pdf', bbox_inches='tight')
plt.show()
print('Saved Fig_VHI_drought_maps.png / .pdf')
```
